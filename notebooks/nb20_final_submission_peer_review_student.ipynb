## 4. Peer Review Rubric

### Why Peer Review?

In industry, no analytics deliverable ships without review. Before HomeValue's pricing report reaches the CEO, a senior analyst reviews the methodology. Before MedScreen's screening tool goes live at a partner hospital, a colleague checks reproducibility and flags blind spots. Peer review catches what self-audit misses — and the reviewer learns as much as the author.

> **"You learn as much from reviewing others' work as from your own."**  
> Good feedback requires critical thinking and makes you a better analyst.

### 4.1 Review Principles

**Be useful, not nice:**
- Focus on specific, actionable feedback
- Point out both strengths and weaknesses
- Suggest concrete improvements

**Be professional:**
- Critique the work, not the person
- Use "I notice" instead of "You did wrong"
- Balance critical feedback with positive observations

**Be thorough:**
- Check technical correctness
- Evaluate communication clarity
- Consider business relevance — would the CEO or oncologist act on these results?

**Why this matters:** Scoring with a rubric removes subjectivity and ensures every reviewer evaluates the same criteria. Use the descriptors to anchor your scores — not gut feeling.

## 6. Wrap-Up: Key Takeaways (and Course Closeout)

### What We Learned Today:

1. **Self-audit before submitting — every time.** Restart the kernel, run all, check the manifest, re-read the output. Five minutes of self-audit catches most of what a reviewer would flag in the first ten minutes of review.
2. **The artifact manifest is the index to everything.** Links to the notebook, the slides, the video, the model card, the submission.csv — one document, every artifact. A reviewer who cannot find your submission in 30 seconds has already started grading down.
3. **Peer review is specific or it is useless.** Three concrete strengths, three concrete suggestions, no vague "analysis seems shallow." The 3-and-3 structure improves the work without damaging the relationship.
4. **Postmortems are honest or they are worthless.** "Here's what I got wrong and what I'd do differently" is the one-sentence posture. A postmortem that claims nothing went wrong is read as either dishonest or un-reflective — both worse than admitting mistakes.

Today ties the bow.

### Critical Rules (the ones that carry out of this course):

> **"Never trust a single split — CV with 95% CI, every time."**

> **"Overlapping CIs mean no winner — pick the simpler model."**

> **"If a preprocessing step uses any data statistic, it belongs inside the Pipeline."**

> **"Slice metrics by group — always, every time, no exceptions."**
