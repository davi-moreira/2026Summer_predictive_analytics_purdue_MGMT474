## 3. Context MattersEvery analysis has a goal and an audience. Three rules follow:- **Separate exploration from final analysis.** Resist the temptation to show everything you did. The reader did not ask for your debugging history.- **Adapt the report to the audience.** Decision-makers rarely care about execution details. ## 4. Visualization Derives From DataBefore you choose a chart, ask three questions about the data behind it:1. **What type of data is this?** Categorical, ordered, continuous, time-indexed?2. **How many dimensions are you encoding?** One? Two? Three or more?3. **Where will the chart be consumed?** Most reports — and every conference poster — are consumed in 2D from three feet away. Encoding more than two dimensions on a static page usually confuses the reader.And one rule that overrides all three: **be careful with scales.** A bad axis can flip the conclusion of a chart without changing the underlying data.

### 4.1 Use Graphs Instead of Tables

<center><img src="figures/table_vs_plot.png" width="60%"/><br><em>The table technically contains the answer. The plot reveals it at a glance.</em></center>

<center><img src="figures/table_vs_plot_02.png" width="60%"/><br><em>Same numbers, two encodings — the chart wins on every reader.</em></center>

When the relationship between numbers is comparative or trend-based, a chart usually beats a table. Reserve tables for cases where readers need to **look up specific values**, not interpret a pattern. *(Source: Kastellec & Leoni, 2007 — "Using Graphs Instead of Tables in Political Science.")*---

### 4.2 Scales Can Be Misleading

<center><img src="figures/escala-fail.PNG" width="60%"/><br><em>Truncated y-axes turn a 3-point swing into a visual landslide.</em></center>

For poster figures, default to **axes that start at zero** for any quantity where zero is a meaningful reference (counts, percentages, currency). When a non-zero baseline is justified — for example, a tightly clustered time series — say so explicitly in the caption.---

### 4.3 Avoid Dual Axes

<center><img src="figures/eixo-duplo.png" width="55%"/><br><em>Two y-axes invite the eye to read causation that the data does not support.</em></center>

### 4.4 …And Triple Axes

<center><img src="figures/eixo-triplo.png" width="60%"/><br><em>If two axes were a stretch, three are a magic trick.</em></center>

Dual-axis charts force the reader to mentally align two scales. When the alignment is arbitrary — and it almost always is — the chart hides as much as it reveals. Prefer **two side-by-side panels** with a shared x-axis, or **standardize both series** (e.g., to z-scores or percent-change-from-baseline) and overlay them on a single shared axis.---

### 4.5 Fewer Pie Charts

<center><img src="figures/pie-vs-bar.png" width="70%"/><br><em>The pie chart asks the reader to compare angles. The bar chart asks them to compare lengths. Lengths win.</em></center>

### 4.6 What?!

<center><img src="figures/steve-jobs-pie.jpg" width="60%"/><br><em>The slices add up to 130%. Yes, this shipped.</em></center>

**What this teaches:** when slice counts are *summed by hand* from labels rather than computed from data, totals can drift past 100% without anyone noticing. The pie format hides the error — three-dimensional rotation and shading make precise area comparison impossible, so the eye accepts the slices as plausible. The fix is structural, not cosmetic: replace the pie with a horizontal bar chart whose lengths are computed directly from the data, and let the reader see the totals.


### 4.7 Oof

<center><img src="figures/escala-fail.PNG" width="55%"/><br><em>Truncated y-axis turns a 2 percentage-point change into an apparent doubling.</em></center>

**What this teaches:** a y-axis that does not start at zero exaggerates differences. The two bars look like one is roughly twice the other, but the underlying values differ by a few percentage points. For a poster, the rule is simple: **bar charts always start at zero**, and line charts only deviate when you are showing change-from-baseline (and label the axis accordingly). Truncated axes are not a stylistic choice; they are misleading.


A question that often comes up here is *"is the pie chart really banned?"* Not banned — but reserve it for the rare case of **two or three categories that sum to a meaningful whole**, where the headline is the overall split rather than a fine comparison between slices. For anything else (especially anything ordered or anything with more than three categories) a horizontal bar chart will outperform a pie chart on every reader test we have ever run in this course.---

## 5. Less Is More — Separate Signal From Noise

The more visual elements your chart contains, the higher the **cognitive load** on the reader. Every line, label, gridline, and decoration is a small tax. Your job is to lower that tax.

This is what Edward Tufte called the **data-ink ratio**:

$$\text{Data-Ink Ratio} = \frac{\text{Data-Ink}}{\text{Total ink used to print the graphic}}$$

The numerator is the ink that *encodes data*. The denominator is everything you printed. Your goal is to push the ratio toward 1.0 by deleting any ink that does not encode data — and only that ink.

The next nine sub-sections walk a single bar chart through a **seven-step cleanup**, removing one piece of non-data ink at a time. Watch what happens at each step: the underlying numbers never change, but the chart's signal grows louder as the noise around it falls away. By the time we reach the "before and after" panel in §5.9, the same data tells a much clearer story.

**Synthesizing the seven steps.** The cleanup did three things, in order: (i) deleted ink that encoded nothing — borders and gridlines; (ii) replaced redundant ink — points behind bars, axes that repeated the labels; (iii) chose colors and labels that earned their place. Each step is small. Together they double the data-ink ratio without losing a single number.

A question that often comes up here is *"isn't a stripped-down chart cold or boring?"* It is the opposite. The more non-data ink you remove, the more visually loud the **remaining** ink becomes. The before/after pair above does not lose information — it amplifies the signal that was already there. The chart on the right is what your poster should look like.

## 6. Hierarchy Among DataOnce you have a clean chart, you have to choose **what the reader looks at first**. That is hierarchy. The eye is not democratic; it lands on whatever is loudest. Your job is to make the loudest thing the *right* thing.

Hierarchy works because the reader's eye filters before it counts. The same idea drives every good poster figure: the headline number should be visually **louder** than every other element, by a factor large enough that you can spot it from across a room.---

A question that often comes up here is *"how many colors should a poster figure have?"* For headline figures: **two**. One muted base color (greys, light blues) for everything that is context, and **one** accent color for the number you want the reader to walk away with. If a chart has more than three colors, ask whether each one is doing structural work — and if it isn't, pick the first three and grey out the rest.---

### 8.6 Same Story, Time-Series Edition

The same six-pass approach works on a time series. The next nine panels (§8.6.1 – §8.6.9) walk a single line chart from the matplotlib default through a fully annotated, story-telling figure. **Notice what changes and what does not**: the underlying data is identical from panel one to panel nine. Only the *annotation layer* — title, color, labels, gridlines, anchor numbers — changes. That is the lesson: most "good" charts you admire on Bloomberg or in *The Economist* are **default plots plus a thoughtful annotation layer**, not exotic chart types.


<center><img src="figures/hist-1.PNG" width="65%"/><br><em>Step 1 — default time series.</em></center>

This is what `plt.plot(df["date"], df["y"])` gives you out of the box: line, x and y ticks, no title, no story. It is the *raw evidence* — useful for the analyst's notebook, not for the audience.


<center><img src="figures/hist-2.PNG" width="65%"/><br><em>Step 2 — title states the conclusion.</em></center>

The first annotation pass replaces the missing title with the *conclusion* the figure supports — not the topic ("Sales over time") but the verdict ("Sales doubled in Q4"). A title that states the conclusion lets the reader know what to look for before they look.


<center><img src="figures/hist-3.PNG" width="65%"/><br><em>Step 3 — accent color on the series of interest.</em></center>

When more than one line is on the chart, accent color isolates the series the title refers to. Everything else fades to grey — present as context, suppressed as story. This is the hierarchy principle from §6 applied to color, not size.


<center><img src="figures/hist-4.PNG" width="65%"/><br><em>Step 4 — annotate the inflection point.</em></center>

A single text annotation at the inflection point ("policy change", "product launch", "incident") explains *why* the line moved when it did. Without that annotation, the reader has to guess; with it, the chart tells a complete story.


<center><img src="figures/hist-5.PNG" width="65%"/><br><em>Step 5 — drop the legend, label the line directly.</em></center>

A legend forces the reader's eye to bounce between the chart and the legend box. Direct labels at the right end of each line keep the eye moving left-to-right and remove the cognitive tax of "which color was Series A again?".


<center><img src="figures/hist-6.PNG" width="65%"/><br><em>Step 6 — trim gridlines that no longer earn their ink.</em></center>

Gridlines help only when readers need to read precise values. After the inflection-point annotation, they do not — the reader's job is now to follow the *story*, not interpolate exact y-values. Drop them.


<center><img src="figures/hist-7.PNG" width="65%"/><br><em>Step 7 — add the contextual number that anchors the story.</em></center>

A single oversized number ("**+82%**") next to the headline anchors the chart numerically. The reader walks away with one fact they can repeat — and that fact survives the moment they look up from the poster.


<center><img src="figures/hist-8.PNG" width="65%"/><br><em>Step 8 — close the gap between title and data.</em></center>

Tight vertical spacing between the title, subtitle, and chart area pulls the reader's eye into the figure. Loose spacing scatters the components into separate visual regions; tight spacing presents the chart as one integrated object.


<center><img src="figures/hist-9.PNG" width="85%"/><br><em>Before and after — same numbers, transformed reading experience.</em></center>

**The synthesis.** Eight passes turned a default matplotlib line into a poster figure that delivers its conclusion in under three seconds. None of the passes changed the data or required a fancier chart type. They added a title that states the verdict, accent color on the series that matters, an annotation explaining the inflection, direct labels in place of a legend, and one anchor number the reader will repeat after walking away. This is the entire content of "good data communication" on a poster — repeated for every figure you put on the wall.


A question that often comes up here is *"do I have to walk every poster figure through nine annotation passes?"* No — but every headline figure should pass the **'one-sentence summary'** test: cover up everything but the chart and ask a teammate to state the conclusion in one sentence. If they can, the figure is doing its job. If they cannot, you are missing one of the six principles — most often hierarchy or annotation.---

## 12. Planning Your Poster — Visual HierarchyA poster has hierarchy at **three layers**: the layout (what's where), the typography (what's loudest), and the content (what's most important). All three have to agree.

<table style="width:100%; border:none;"><tr><td style="width:40%; text-align:center; border:none;"><img src="figures/poster_template.png" width="90%"/></td><td style="width:60%; vertical-align:middle; border:none; padding-left:20px;"><ul><li>Provide a visually engaging, <strong>easy-to-navigate</strong> summary.</li><li><strong>Always lay out in columns</strong> — typically three for a 36×48 inch poster.</li><li><strong>Headings:</strong> Large and bold, in a consistent hierarchy that guides the reader.</li><li><strong>Color and contrast:</strong> Pick a simple palette (one accent + neutrals) that highlights the main points without competing with your figures.</li><li><strong>Font size:</strong> Body text should be legible from <strong>three feet away</strong> — that is the test, not "looks fine on my laptop screen".</li><li><strong>Flow:</strong> Logical reading order, top-left to bottom-right.</li></ul></td></tr></table>

## 14. Planning Your Poster — Organizing Your ContentEvery research poster, regardless of field, follows roughly the same eleven-section architecture. Adapt this list to your project — most groups will compress sections 6–8 and expand section 9.

<table><tr><th>#</th><th>Section</th><th>What goes here</th></tr><tr><td>1</td><td><strong>Title and Authors</strong></td><td>Clear, concise, visible from across the room. Plus the four group members + course + advisor.</td></tr><tr><td>2</td><td><strong>Abstract / Introduction</strong></td><td>One paragraph: <em>why the problem matters + what the main result is</em>.</td></tr><tr><td>3</td><td><strong>Business Problem</strong></td><td>What is the project goal in stakeholder language?</td></tr><tr><td>4</td><td><strong>Analytics Problem Framing</strong></td><td>What you aim to predict or explain — and why this is the right framing.</td></tr><tr><td>5</td><td><strong>Research Question</strong></td><td>One sentence summarizing your project as a question you will answer.</td></tr><tr><td>6</td><td><strong>Data and EDA</strong></td><td>Key variables, sample size, key descriptive plot.</td></tr><tr><td>7</td><td><strong>Methodology</strong></td><td>Modeling approach focused on the response variable. Include the leakage-safe pipeline diagram.</td></tr><tr><td>8</td><td><strong>Model Building and Evaluation</strong></td><td>Cross-validation results, baseline comparison, the locked-test-set verdict.</td></tr><tr><td>9</td><td><strong>Business Implication</strong></td><td>Validation of the model in stakeholder terms — what decision changes.</td></tr><tr><td>10</td><td><strong>Conclusion and Future Work</strong></td><td>Wrap-up + potential next steps.</td></tr><tr><td>11</td><td><strong>Acknowledgments and References</strong></td><td>Recognize collaborators and cite data sources / libraries.</td></tr></table>A question that often comes up here is *"can I drop a section if my project doesn't need it?"* Yes — if you can defend the omission to a judge in one sentence. The most common defensible cut is collapsing **Research Question** (section 5) into **Analytics Problem Framing** (section 4) when the two phrasings are nearly identical. The most common indefensible cut is dropping **Acknowledgments and References** because you "ran out of room" — that signals a layout problem, not a content problem.---

## 15. Designing for Predictive Analytics — The Special CaseMost poster-design advice is generic. This section is the predictive-analytics-specific layer — the parts of the poster judges will look at *because your project is a modeling project*.

<table style="width:100%; border:none;"><tr><td style="width:40%; text-align:center; border:none;"><img src="figures/poster_template.png" width="90%"/></td><td style="width:60%; vertical-align:middle; border:none; padding-left:20px"><p><strong>Emphasize the predictive component</strong></p><ul><li><strong>Model choice:</strong> State explicitly whether you used logistic regression, random forest, gradient boosting, etc. Do not assume the reader will guess.</li><li><strong>Metrics:</strong> Include accuracy, RMSE, ROC-AUC, or whichever metric was the headline of your evaluation — with the 95% CI from your cross-validation work.</li><li><strong>Model interpretation:</strong> Highlight the top three to five features that drove the prediction. A bar chart of feature importance is usually the right shape.</li></ul><p><strong>Data visualization tips</strong></p><ul><li>Label every <strong>axis and legend</strong>; drop the legend when direct labels would do.</li><li>Show <strong>before-and-after comparisons</strong> if you performed substantive feature engineering.</li><li>Visually emphasize the <strong>training-vs-testing split</strong> (or cross-validation diagram) so a non-expert reader sees that you didn't evaluate on data the model had seen.</li></ul></td></tr></table>

## 18. Effective Figures and Tables on a PosterEverything you learned in Sections 3–8 of this notebook applies here. Two reminders specifically for the poster medium:

<table style="width:100%; border:none;"><tr><td style="width:40%; text-align:center; border:none;"><img src="figures/poster_template.png" width="90%"/></td><td style="width:60%; vertical-align:middle; border:none; padding-left:20px;"><p><strong>Charts and graphs</strong></p><ul><li>Label <strong>axes</strong>; include legends only if direct labels are not feasible.</li><li>Highlight the data that supports your <strong>main argument</strong> — accent color on the headline series, muted greys on the rest.</li><li>Provide <strong>short, descriptive captions</strong> beneath each figure.
